```python
import math

def calculate_u(u, n):
    temp = u
    for i in range(1, n):
        temp = temp * (u - i)
    return temp

def newton_forward_interpolation():
    n = 4
    x = [45, 50, 55, 60]
    y = [[0 for _ in range(n)] for _ in range(n)]
    
    y[0][0] = 0.7071
    y[1][0] = 0.7660
    y[2][0] = 0.8192
    y[3][0] = 0.8660

    for i in range(1, n):
        for j in range(n - i):
            y[j][i] = y[j + 1][i - 1] - y[j][i - 1]

    value = 52
    
    sum_val = y[0][0]
    u = (value - x[0]) / (x[1] - x[0])
    
    for i in range(1, n):
        sum_val = sum_val + (calculate_u(u, i) * y[0][i]) / math.factorial(i)

    print(f"Interpolated value at {value} is {round(sum_val, 6)}")

newton_forward_interpolation()
```
